# Smart Student Performance Prediction & Academic Risk Analytics
## Exploratory Data Analysis & Early Warning Validation
**Dataset:** UCI Student Performance Dataset (Cortez & Silva, 2008)
**Target Variable:** `at_risk` (Binary Classification: $G3 < 10$ on statutory 0-20 scale)
**Scope:** Preventing Academic Attrition using Mid-term Formative Evaluation ($G1$) without Data Leakage

```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 150

df = pd.read_csv('../data/processed/student_performance_cleaned.csv')
print(f'Total records: {len(df):,}')
print(f'Unique students: {df["student_id"].nunique():,}')
df.head(3)
```

### 1. Data Integrity and Target Class Distribution

```
print('Missing values audit: Total missing =', df.isnull().sum().sum())
print('\nTarget Class Breakdown:')
print(df['at_risk'].value_counts())
print('\nTarget Class Ratio:')
print(df['at_risk'].value_counts(normalize=True).apply(lambda x: f'{x:.2%}'))
print('\nCourse-specific At-Risk Rate:')
print(df.groupby('course')['at_risk'].agg(['count', 'mean']).rename(columns={'mean': 'at_risk_rate'}))
```

### 2. Numerical Features Summary

```
num_cols = ['age', 'Medu', 'Fedu', 'traveltime', 'studytime', 'failures', 'absences', 'G1', 'G3']
df[num_cols].describe().round(2)
```

### 3. Key Relationships with Academic Risk

```
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(x='at_risk', y='G1', data=df, ax=axes[0], palette=['#059669', '#dc2626'], hue='at_risk', legend=False)
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(['Satisfactory (0)', 'At Risk (1)'])
axes[0].set_title('Mid-term Grade (G1) by At-Risk Outcome')

fail_risk = df.groupby('failures')['at_risk'].mean() * 100
axes[1].bar([str(i) for i in fail_risk.index], fail_risk.values, color='#dc2626', edgecolor='black')
axes[1].set_title('At-Risk Rate (%) by Prior Past Failures')
axes[1].set_ylabel('At Risk (%)')
plt.tight_layout()
plt.show()
```
